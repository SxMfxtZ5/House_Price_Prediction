# Import Libraries

```python
import pandas as pd
import datetime
```

# Load in data

```python
data = pd.read_csv('data/raw/PPR-ALL.csv', encoding='ISO-8859-1')
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

### Convert date to yyyy-mm-dd based on ISO 8601 standards

```python
data["Date of Sale (dd/mm/yyyy)"] = pd.to_datetime(
    data["Date of Sale (dd/mm/yyyy)"],
    format="%d/%m/%Y",
    errors='coerce'
)

data.head()
```

### Fix address column

```python
address = data["Address"].str.split(',').apply(lambda lst: [i.strip() for i in lst])

data["street"]   = address.str[0]
data["locality"] = address.apply(lambda lst: ",".join(lst[1:-1]) if len(lst) > 2 else None)
data.head()
```

### Dropping eircode column due to high number of null values. May introduce noise.

```python
data.drop(columns=["Eircode"])
```
